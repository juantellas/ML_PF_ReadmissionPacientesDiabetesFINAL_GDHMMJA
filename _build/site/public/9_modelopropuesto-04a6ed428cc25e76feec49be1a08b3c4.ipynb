### **9.1.1 Paper A - “Comparison of machine learning models for predicting 30-day readmission rates for patients with diabetes”**


El artículo **“Comparison of machine learning models for predicting 30-day readmission rates for patients with diabetes”** analiza la capacidad predictiva de distintos algoritmos de aprendizaje automático para identificar pacientes diabéticos con riesgo de reingresar al hospital en los 30 días posteriores al alta. Los autores utilizan el conjunto de datos *Diabetes 130-US Hospitals*, y aplican un proceso riguroso de preprocesamiento que incluye balanceo de clases mediante SMOTE, codificación de variables categóricas, estandarización y selección de características usando el método metaheurístico **Grey Wolf Optimizer (GWO)**. Con estas bases, construyen un marco comparativo entre **diez modelos de machine learning tradicional** —SVM con kernel lineal y RBF, Random Forest, XGBoost, Logistic Regression, KNN, Naïve Bayes, AdaBoost, Decision Tree y MLP— además de un modelo de aprendizaje profundo, **LSTM**, con el propósito de evaluar cuál enfoque ofrece un desempeño más fiable en este contexto clínico.

Los resultados muestran que los modelos de **Machine Learning tradicional superan al modelo de Deep Learning** en este problema específico. En particular, **Random Forest y XGBoost** alcanzan los mejores desempeños, con valores de F1-score cercanos a 0.83–0.84 y una exactitud aproximada del 88 %, destacándose por su estabilidad y capacidad para manejar relaciones no lineales en los datos. Por el contrario, el modelo **LSTM**, pese a ser el único algoritmo de deep learning incluido, obtiene métricas sustancialmente menores (F1 alrededor de 0.79 y AUROC en torno a 0.61), lo que evidencia que, en un entorno con variables tabulares y una estructura temporal limitada, los modelos clásicos pueden ser más eficaces y eficientes.

En conjunto, el estudio concluye que técnicas tradicionales bien optimizadas —especialmente los métodos de ensamblado como Random Forest y XGBoost— ofrecen un rendimiento superior al del modelo LSTM en la predicción del reingreso a 30 días. De esta forma, el artículo demuestra que la complejidad del modelo no necesariamente garantiza mejores resultados, y subraya la importancia de una adecuada ingeniería de características y estrategias de balanceo para obtener predicciones confiables en el ámbito clínico.

DOI: https://dx.doi.org/10.21037/jmai-24-70

### **9.1.2 Paper B - “A comparative study of software defect binomial classification prediction models based on machine learning”**


El artículo **“A comparative study of software defect binomial classification prediction models based on machine learning”** (Tao et al., 2024) presenta un análisis amplio y sistemático de trece modelos de aprendizaje automático aplicados al problema de clasificación binaria de defectos de software. Aunque el dominio es diferente al de la salud, el estudio es metodológicamente pertinente porque evalúa modelos ampliamente utilizados en tareas de predicción sobre datos tabulares. Los autores trabajan con múltiples conjuntos de datos estandarizados en ingeniería de software y aplican un protocolo uniforme de preprocesamiento, lo que permite comparar de manera fiable el desempeño de modelos clásicos como **Logistic Regression, Naïve Bayes, KNN, Decision Tree y SVM**, junto con métodos más avanzados basados en ensamblado, tales como **Random Forest, AdaBoost, Extra Trees, Bagging, Gradient Boosting, CatBoost, LightGBM y XGBoost**. Esta variedad de algoritmos ofrece una perspectiva clara sobre el comportamiento relativo de distintas familias de modelos.

Los resultados del estudio indican que los métodos de ensamblado basados en árboles son los más efectivos para este tipo de datos. En particular, modelos como **CatBoost, LightGBM y Extra Trees** alcanzan los valores más altos de precisión, estabilidad y capacidad de generalización entre los conjuntos de datos evaluados. Los autores destacan que estos algoritmos superan consistentemente a modelos más simples como Naïve Bayes, KNN o Logistic Regression, especialmente en entornos con interacciones complejas entre variables. Asimismo, modelos como **XGBoost y Gradient Boosting** también obtienen un rendimiento competitivo, reafirmando la eficacia de los enfoques basados en boosting. En conjunto, las conclusiones del artículo subrayan que la combinación de árboles de decisión con estrategias de mejora iterativa constituye una de las alternativas más robustas para la predicción en dominios tabulares, aportando evidencia relevante para comparar y contextualizar el desempeño de modelos similares en otras áreas de aplicación.


DOI: https://doi.org/10.1007/s11219-024-09683-3
